##### We treat y1 = dfPG["target_tre200h0_plus12h"]

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y1, test_size = .2, random_state = 12)
```

##### We want to find the best coefficient with lasso
* we use a pipeline to have better performance
* we use a standard scaler to have not too much variance between the values display on a plot x y

```python
from sklearn.linear_model import Lasso
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
lasso = Lasso(alpha = 0.1, fit_intercept = True)

lasso_pipe = Pipeline([
    ('scaler', StandardScaler()),
    ('lasso', lasso)
])
```

```python
lasso_pipe.fit(X_train, y_train)
```

```python
from sklearn.metrics import root_mean_squared_error, r2_score

print(f"Variance de y : {np.var(y1)}" )
print(f"Ecart-type de y : {np.std(y1)}")
y_hat = lasso_pipe.predict(X_train)
print(f"MSRE_train : {root_mean_squared_error(y_train, y_hat)}")
y_hat_test = lasso_pipe.predict(X_test)
print(f"MSRE_test : {root_mean_squared_error(y_test, y_hat_test)}")
```

**En regression on utilise R2 score**
* En régression, on ne parle pas d'accuracy comme en classification. On utilise plutôt le R² score, qui indique la proportion de la variance expliquée par le modèle.

```python
lasso_gride = np.linspace(1, 1e6) # tunning parameters to find the main coefficient
```

```python
coefs = []
for alpha in lasso_gride:
    lasso_pipe.set_params(lasso__alpha=alpha, lasso__fit_intercept=True)
    lasso_pipe.fit(X_train, y_train)
    coefs.append(lasso_pipe.named_steps['lasso'].coef_)
```

```python
def affichage_lasso_Coef_with_alpha(grid_alpha, X_train, y_train):
    coef = []
    for alpha in grid_alpha:
        lp =lasso_pipe.set_params(lasso__alpha=alpha, lasso__fit_intercept=True).fit(X_train, y_train)
        coef.append(lp.named_steps['lasso'].coef_)

    coef_lp = np.array(coef)

    plt.figure(figsize=(10,6))
    for i in range(coef_lp.shape[1]):
        plt.plot(grid_alpha, coef_lp[:, i], marker="o", label=f"Feature {i}")
    plt.xscale("log")
    plt.xlabel("Alpha (log scale)")
    plt.ylabel("Coefficient value")
    plt.title("Évolution des coefficients lasso selon alpha")
    plt.legend(X_train.columns, loc='upper right', bbox_to_anchor = (2.5, 1))
    plt.legend(bbox_to_anchor=(1, 1), loc="upper left")
    plt.show()
```

```python
affichage_lasso_Coef_with_alpha(lasso_gride, X_train, y_train)
```

###  CV for Lasso

```python
from sklearn.model_selection import KFold
folds = KFold(n_splits=10 , shuffle=True, random_state=42)
```

```python
#lasso_gride = np.linspace(1e-2, 1e6)
#lasso_gride = 10**np.linspace(0, 6, num=30)
lasso_gride = np.geomspace(1, 1e6, num=30)
dictAlpha_lasso = {
    "lasso__alpha": lasso_gride,
}
dictAlpha_lasso['lasso__alpha']
params_lasso = {"lasso__alpha" : lasso_gride}
```

```python
from sklearn.model_selection import GridSearchCV
gscv_lasso = GridSearchCV(estimator=lasso_pipe, param_grid=dictAlpha_lasso, scoring="neg_mean_squared_error", cv=folds)
```

Run the cross-validation by calling the `fit` method of the **cross-validation object** that you created at the point before.

```python
lassoCV = gscv_lasso.fit(X_train, y_train)
```

We want to plot the cross-validation error against different values of the penalty parameter

```python
df_lasso_cv_results_ = pd.DataFrame(lassoCV.cv_results_)
df_lasso_cv_results_.info()
df_lasso_cv_results_.head(5)
```

Run the cross-validation by calling the `fit` method of the **cross-validation object** that you created at the point before.

```python
lassoCV.fit(X_train, y_train)
```

We want to plot the cross-validation error against different values of the penalty parameter.

```python
# Choose best model that minimizes the cv_err
mean_scores = -lassoCV.cv_results_["mean_test_score"]
se_scores = lassoCV.cv_results_["std_test_score"] / np.sqrt(lassoCV.n_splits_)
alphas = lassoCV.cv_results_["param_lasso__alpha"].data

best_index = np.argmin(mean_scores)
min_alpha_lasso = alphas[best_index]

threshold_lasso = mean_scores[best_index] + se_scores[best_index]
one_se_rule_alpha_lasso = np.max(alphas[mean_scores <= threshold_lasso])

print("Minimum alpha:", min_alpha_lasso)
print("1-SD alpha:", one_se_rule_alpha_lasso)
print("Best score for lasso:", np.sqrt(np.min(mean_scores)))
```

```python
plt.errorbar(x=lasso_gride, y=mean_scores, yerr = se_scores, fmt='o', capsize=3)

plt.axvline(min_alpha_lasso, ls='dotted', color="grey")
plt.axvline(one_se_rule_alpha_lasso, ls='dotted', color="grey")
plt.axhline(threshold_lasso, ls='dotted', color="grey")

plt.title("Lasso regressor CV error")
plt.xlabel('log(alpha)')
plt.ylabel('Mean Squared Error')
plt.xscale('log')
plt.show()
```

```python
def preTraitement(dfPG):
    dfPG = pd.concat([dfPG, pd.get_dummies(dfPG['season'])], axis=1)
    dfPG = dfPG.drop(columns=['season'])
    dfPG = dfPG.dropna()
    #x = dfPG.drop(columnsNotWanted,axis=1)
    x = dfPG.drop(["target_tre200h0_plus12h", "target_tre200h0_plus24h", "target_tre200h0_plus48h"], axis=1)
    y1 = dfPG["target_tre200h0_plus12h"]
    y2 = dfPG["target_tre200h0_plus24h"]
    y3 = dfPG["target_tre200h0_plus48h"]
    return {
        'X': x,
        'y1': y1,
        'y2': y2,
        'y3': y3
    }
```

```python
def preTraitement_test(df_test):
    df_test = pd.concat([df_test, pd.get_dummies(df_test['season'])], axis=1)
    df_test = df_test.drop(columns=['season', 'Id'])
    #df_test = df_test.drop(columnsNotWanted2,axis=1)
    df_test = df_test.dropna()
    #X = df_test.drop(columnsNotWanted2,axis=1)
    return {
        'X':df_test
    }
```
